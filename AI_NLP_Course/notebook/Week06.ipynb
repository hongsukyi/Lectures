## Practice Problem

**Scenario:** Using the same offline `sklearn.datasets.load_digits`
dataset (8x8 grayscale digit images, 10 classes), build and compare an
MLP and a CNN, using the **same method** as this lecture.

1. Load the digits dataset, split it into train (60%), validation
   (20%), and test (20%) with `train_test_split(..., stratify=...)`.
2. Build `train_loader`, `val_loader`, `test_loader` with
   `DataLoader(..., batch_size=32)`.
3. Build the same two model structures:
   - MLP: `Flatten -> Linear(64, 64) -> ReLU -> Linear(64, 10)`
   - CNN: `Conv2d(1, 8, 3, padding=1) -> ReLU -> MaxPool2d(2) ->
     Conv2d(8, 16, 3, padding=1) -> ReLU -> AdaptiveAvgPool2d((2, 2))
     -> Flatten -> Linear(16*2*2, 10)`
4. Train both for 15 epochs with `CrossEntropyLoss` and `Adam(lr=0.005)`.
5. Compare final test accuracy.
6. Plot the confusion matrix for the CNN.


### Solution

```python
import numpy as np
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
import torch.optim as optim
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, confusion_matrix

torch.manual_seed(42)
np.random.seed(42)
```

```python
from torch.utils.data import TensorDataset, Subset, DataLoader
from sklearn.metrics import ConfusionMatrixDisplay
import pandas as pd
from IPython.display import display  # Colab/Jupyter 환경이 아니면 print(test_summary_p.round(4))로 대체
```

```python
# ----- Data -----
digits_p = load_digits()
images_p = torch.tensor(digits_p.images, dtype=torch.float32).unsqueeze(1) / 16.0
labels_p = torch.tensor(digits_p.target, dtype=torch.long)

all_indices_p = list(range(len(images_p)))
train_val_indices_p, test_indices_p = train_test_split(
    all_indices_p, test_size=0.20, random_state=42, stratify=digits_p.target,
)
train_indices_p, val_indices_p = train_test_split(
    train_val_indices_p, test_size=0.25, random_state=42,
    stratify=digits_p.target[train_val_indices_p],
)

full_dataset_p = TensorDataset(images_p, labels_p)
train_dataset_p = Subset(full_dataset_p, train_indices_p)
val_dataset_p = Subset(full_dataset_p, val_indices_p)
test_dataset_p = Subset(full_dataset_p, test_indices_p)

class_names_p = [str(i) for i in range(10)]
print("Train/val/test samples:", len(train_dataset_p), len(val_dataset_p), len(test_dataset_p))

train_loader_p = DataLoader(train_dataset_p, batch_size=32, shuffle=True)
val_loader_p = DataLoader(val_dataset_p, batch_size=32, shuffle=False)
test_loader_p = DataLoader(test_dataset_p, batch_size=32, shuffle=False)

# ----- Models -----
torch.manual_seed(42)
mlp_model_p = nn.Sequential(
    nn.Flatten(),
    nn.Linear(64, 64),
    nn.ReLU(),
    nn.Linear(64, 10),
)

cnn_model_p = nn.Sequential(
    nn.Conv2d(1, 8, kernel_size=3, padding=1),
    nn.ReLU(),
    nn.MaxPool2d(kernel_size=2),
    nn.Conv2d(8, 16, kernel_size=3, padding=1),
    nn.ReLU(),
    nn.AdaptiveAvgPool2d((2, 2)),
    nn.Flatten(),
    nn.Linear(16 * 2 * 2, 10),
)

loss_fn_p = nn.CrossEntropyLoss()
models_p = {"MLP": mlp_model_p, "CNN": cnn_model_p}
epochs_p = 15
histories_p = {}

for model_name, current_model in models_p.items():
    optimizer_p = torch.optim.Adam(current_model.parameters(), lr=0.005)
    val_accuracy_history_p = []

    for epoch in range(epochs_p):
        current_model.train()
        for batch_images, batch_labels in train_loader_p:
            optimizer_p.zero_grad()
            loss = loss_fn_p(current_model(batch_images), batch_labels)
            loss.backward()
            optimizer_p.step()

        current_model.eval()
        val_correct, val_count = 0, 0
        with torch.no_grad():
            for batch_images, batch_labels in val_loader_p:
                logits = current_model(batch_images)
                val_correct += (logits.argmax(dim=1) == batch_labels).sum().item()
                val_count += len(batch_images)
        val_accuracy_history_p.append(val_correct / val_count)

    histories_p[model_name] = val_accuracy_history_p
    print(f"{model_name}: final validation accuracy = {val_accuracy_history_p[-1]:.3f}")

# ----- Test comparison -----
test_summary_p = []
test_predictions_by_model_p = {}
for model_name, current_model in models_p.items():
    current_model.eval()
    all_predictions, all_targets = [], []
    with torch.no_grad():
        for batch_images, batch_labels in test_loader_p:
            predictions = current_model(batch_images).argmax(dim=1)
            all_predictions.append(predictions)
            all_targets.append(batch_labels)
    all_predictions = torch.cat(all_predictions)
    all_targets = torch.cat(all_targets)
    test_predictions_by_model_p[model_name] = all_predictions
    test_summary_p.append({"model": model_name, "test_accuracy": accuracy_score(all_targets.numpy(), all_predictions.numpy())})

test_summary_p = pd.DataFrame(test_summary_p)
display(test_summary_p.round(4))

plt.figure(figsize=(5, 3))
plt.bar(test_summary_p["model"], test_summary_p["test_accuracy"])
plt.ylim(0, 1)
plt.ylabel("Test accuracy")
plt.title("Practice: Final Test Accuracy")
plt.grid(False)
plt.show()

# ----- Confusion matrix for the CNN -----
cnn_targets_p = torch.cat([labels for _, labels in test_loader_p])
cnn_confusion_p = confusion_matrix(cnn_targets_p.numpy(), test_predictions_by_model_p["CNN"].numpy())

fig, ax = plt.subplots(figsize=(5, 3))
ConfusionMatrixDisplay(confusion_matrix=cnn_confusion_p, display_labels=class_names_p).plot(ax=ax)
ax.grid(False)
plt.title("Practice: CNN Confusion Matrix (digits)")
plt.show()
```
