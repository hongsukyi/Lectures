```python
# ==============================
# 3. Matplotlib — Visualization
# ==============================
import matplotlib.pyplot as plt


x = np.linspace(0, 2 * np.pi, 100)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))
ax1.plot(x, np.sin(x), 'r-', label='sin(x)')
ax1.plot(x, np.cos(x), 'b-', label='cos(x)')
ax1.set(title='Line Plot', xlabel='x', ylabel='y')
ax1.legend(); ax1.grid(True, alpha=0.3)

data = np.random.randn(200)
ax2.hist(data, bins=20, color='#6c5ce7', edgecolor='white')
ax2.set(title='Histogram (Normal Distribution)', xlabel='Value', ylabel='Count')
ax2.grid(True, alpha=0.3)

fig.suptitle('Matplotlib Basics', fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout(); plt.show()
```

```python
# ==============================
# 4. Scikit-learn — Simple Classification
# ==============================
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

print("=" * 40)
print("4. Scikit-learn Basics")
print("=" * 40)

iris = load_iris()
X_train, X_test, y_train, y_test = train_test_split(
    iris.data, iris.target, test_size=0.3, random_state=42)

model = LogisticRegression(max_iter=200)
model.fit(X_train, y_train)
y_pred = model.predict(X_test)

print(f"Dataset: IRIS ({iris.data.shape})")
print(f"Classes: {list(iris.target_names)}")
print(f"Accuracy: {accuracy_score(y_test, y_pred):.4f}")
```

```python
# ==============================
# 5. TensorFlow — Simple Neural Network
# ==============================
import tensorflow as tf

print("\n" + "=" * 40)
print("5. TensorFlow Basics")
print("=" * 40)
print(f"TensorFlow version: {tf.__version__}")

model = tf.keras.Sequential([
    tf.keras.layers.Dense(16, activation='relu'),
    tf.keras.layers.Dense(3, activation='softmax')
])
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

model.fit(X_train, y_train, epochs=50, verbose=0)
loss, acc = model.evaluate(X_test, y_test, verbose=0)
print(f"Keras MLP Accuracy: {acc:.4f}")
```

```python
# --- sklearn vs keras comparison ---
fig, ax = plt.subplots(figsize=(6, 4))
names = ['sklearn\nLogistic', 'Keras\nMLP']
accs = [accuracy_score(y_test, y_pred), acc]
bars = ax.bar(names, accs, color=['#0984e3', '#e74c3c'], edgecolor='white', width=0.4)
for b, v in zip(bars, accs):
    ax.text(b.get_x() + b.get_width()/2, v + 0.005, f'{v:.4f}',
            ha='center', fontweight='bold', fontsize=13)
ax.set(title='IRIS Classification — sklearn vs Keras', ylabel='Test Accuracy')
ax.set_ylim(min(accs) - 0.05, max(accs) + 0.04)
ax.grid(axis='y', alpha=0.3)
plt.tight_layout(); plt.show()
```

```python
# ==============================
# 5. PyTorch — Simple Neural Network
# ==============================
import torch
import torch.nn as nn
import torch.optim as optim

print("\n" + "=" * 40)
print("5. PyTorch Basics")
print("=" * 40)
print(f"PyTorch version: {torch.__version__}")

# X_train, y_train, X_test, y_test는 numpy array 또는 torch tensor라고 가정
X_train_t = torch.as_tensor(X_train, dtype=torch.float32)
y_train_t = torch.as_tensor(y_train, dtype=torch.long)
X_test_t = torch.as_tensor(X_test, dtype=torch.float32)
y_test_t = torch.as_tensor(y_test, dtype=torch.long)

class MLP(nn.Module):
    def __init__(self, input_dim, hidden_dim=16, num_classes=3):
        super().__init__()
        self.fc1 = nn.Linear(input_dim, hidden_dim)
        self.relu = nn.ReLU()
        self.fc2 = nn.Linear(hidden_dim, num_classes)

    def forward(self, x):
        x = self.relu(self.fc1(x))
        return self.fc2(x)  # logits, softmax는 손실 함수 안에서 처리

input_dim = X_train_t.shape[1]
model = MLP(input_dim)

criterion = nn.CrossEntropyLoss()  # Keras의 sparse_categorical_crossentropy에 대응
optimizer = optim.Adam(model.parameters())

# ----- Train -----
model.train()
for epoch in range(50):
    optimizer.zero_grad()
    logits = model(X_train_t)
    loss = criterion(logits, y_train_t)
    loss.backward()
    optimizer.step()

# ----- Evaluate -----
model.eval()
with torch.no_grad():
    test_logits = model(X_test_t)
    test_loss = criterion(test_logits, y_test_t).item()
    preds = test_logits.argmax(dim=1)
    acc = (preds == y_test_t).float().mean().item()

print(f"PyTorch MLP Accuracy: {acc:.4f}")
```
