# Lecture 03: Logistic Regression for Binary Classification

## From a linear score to a class probability

This notebook is a **beginner-level PyTorch exercise**.

- Predicts Pass/Fail from two input features.
- Computes a linear score $z$.
- Turns $z$ into a probability with sigmoid.
- Decides the class with a 0.5 threshold.
- Trains with `BCEWithLogitsLoss`.


```python
import torch
from torch import nn
import pandas as pd
import matplotlib.pyplot as plt

# For reproducible results
torch.manual_seed(42)
torch.set_num_threads(1)

print("PyTorch version:", torch.__version__)
```

## 1. Create a Pass/Fail dataset

- Feature 1: study time
- Feature 2: attendance
- Target: Fail = 0, Pass = 1


## Practice Problem

**Scenario:** A university admissions office wants to predict
Admit (1) / Reject (0) from two exam scores.



Using the same method as this lecture (standardize the two
features, build nn.Linear(2, 1), train with BCEWithLogitsLoss and
SGD(lr=0.1) for 1000 epochs), do the following:

1. Train a logistic regression model on this data.
2. Plot the training loss curve.
3. Compute training accuracy and the confusion matrix.
4. Plot the decision boundary.
5. Predict the outcome for a new applicant with Math=63, English=70.
| Math score | English score | Target |
|---|---|---|
| 40 | 45 | 0 |
| 45 | 50 | 0 |
| 50 | 48 | 0 |
| 52 | 55 | 0 |
| 55 | 58 | 0 |
| 58 | 60 | 0 |
| 60 | 65 | 1 |
| 65 | 68 | 0 |
| 68 | 72 | 1 |
| 70 | 75 | 1 |
| 72 | 78 | 1 |
| 75 | 80 | 1 |
| 78 | 85 | 1 |
| 82 | 88 | 1 |
| 85 | 90 | 1 |
| 90 | 92 | 1 |

### Solution

```python
# ----- Data -----
X_p = torch.tensor([
    [40.0, 45.0], [45.0, 50.0], [50.0, 48.0], [52.0, 55.0],
    [55.0, 58.0], [58.0, 60.0], [60.0, 65.0], [65.0, 68.0],
    [68.0, 72.0], [70.0, 75.0], [72.0, 78.0], [75.0, 80.0],
    [78.0, 85.0], [82.0, 88.0], [85.0, 90.0], [90.0, 92.0],    
], dtype=torch.float32)

y_p = torch.tensor([
    0, 0, 0, 0, 0, 0, 1, 0, 1, 1, 1, 1, 1, 1, 1, 1,
], dtype=torch.float32).reshape(-1, 1)

# ----- Standardize, the same way as the lecture -----
feature_mean_p = X_p.mean(dim=0, keepdim=True)
feature_std_p = X_p.std(dim=0, keepdim=True)
X_p_scaled = (X_p - feature_mean_p) / feature_std_p

# ----- Build and train the model -----
torch.manual_seed(42)
model_p = nn.Linear(2, 1)
loss_fn_p = nn.BCEWithLogitsLoss()
optimizer_p = torch.optim.SGD(model_p.parameters(), lr=0.1)   

epochs = 1000
loss_history_p = []

for epoch in range(epochs):
    optimizer_p.zero_grad()
    logits_p = model_p(X_p_scaled)
    loss_p = loss_fn_p(logits_p, y_p)
    loss_p.backward()
    optimizer_p.step()
    loss_history_p.append(loss_p.item())

print("Final loss:", loss_history_p[-1])

# ----- Plot the training loss -----
plt.figure(figsize=(5, 3))
plt.plot(loss_history_p)
plt.xlabel("Epoch")
plt.ylabel("BCE loss")
plt.title("Practice: Training Loss")
plt.grid(False)
plt.show()

# ----- Accuracy and confusion matrix -----
model_p.eval()
with torch.no_grad():
    probabilities_p = torch.sigmoid(model_p(X_p_scaled))
    predictions_p = (probabilities_p >= 0.5).float()
    accuracy_p = (predictions_p == y_p).float().mean()

print("Training accuracy:", accuracy_p.item())

confusion_p = torch.zeros((2, 2), dtype=torch.int64)
for actual, predicted in zip(y_p.squeeze().to(torch.int64), 
                predictions_p.squeeze().to(torch.int64)):
    confusion_p[actual, predicted] += 1

confusion_p_df = pd.DataFrame(
    confusion_p.numpy(),
    index=["Actual 0", "Actual 1"],
    columns=["Predicted 0", "Predicted 1"],
)
display(confusion_p_df)

# ----- Decision boundary -----
math_grid = torch.linspace(X_p[:, 0].min() - 5, X_p[:, 0].max() + 5, 220)
english_grid = torch.linspace(X_p[:, 1].min() - 5, X_p[:, 1].max() + 5, 220) 
M, E = torch.meshgrid(math_grid, english_grid, indexing="xy")

grid_p = torch.stack([M.reshape(-1), E.reshape(-1)], dim=1)
grid_p_scaled = (grid_p - feature_mean_p) / feature_std_p

with torch.no_grad():
    grid_probability_p = torch.sigmoid(
        model_p(grid_p_scaled)).reshape(M.shape)

reject_mask_p = y_p.squeeze() == 0
admit_mask_p = y_p.squeeze() == 1

plt.figure(figsize=(5, 3))
plt.contourf(M.numpy(), E.numpy(), grid_probability_p.numpy(), levels=[0.0, 0.5, 1.0], alpha=0.18)
plt.contour(M.numpy(), E.numpy(), grid_probability_p.numpy(), levels=[0.5], linewidths=2)
plt.scatter(X_p[reject_mask_p, 0], X_p[reject_mask_p, 1], marker="^", s=80, label="Reject (0)")
plt.scatter(X_p[admit_mask_p, 0], X_p[admit_mask_p, 1], marker="o", s=80, label="Admit (1)")
plt.xlabel("Math score")
plt.ylabel("English score")
plt.title("Practice: Decision Boundary")
plt.legend()
plt.grid(False)
plt.show()

# ----- Predict a new applicant -----
new_applicant = torch.tensor([[63.0, 70.0]])
new_applicant_scaled = (new_applicant - feature_mean_p) / feature_std_p

with torch.no_grad():
    new_logit_p = model_p(new_applicant_scaled)
    new_probability_p = torch.sigmoid(new_logit_p)
    new_prediction_p = int(new_probability_p.item() >= 0.5)

print("new applicant:", new_applicant.tolist()[0])
print(f"probability of admit: {new_probability_p.item():.3f}")
print("predicted class:", new_prediction_p)
```
