Beginner PyTorch notebook using only the six provided English sentences.

- One small full-batch sequence-generation example
- Teacher forcing during training and greedy autoregressive decoding during inference

## Practice Problem

**Scenario:** Use the same new four-sentence corpus as the Lecture 12
practice problem:

```python
corpus_p = [
    "Noah walks a dog in the park.",
    "Mia walks a cat in the park.",
    "Emma feeds a bird in the garden.",
    "Liam feeds a fish in the garden.",
]
```

Using the **same method** as this lecture, do the following:

1. Build the compact source sequences from the full target sentences,
   the same way as this lecture: `[subject, base_verb, object, place]`
   (use `base_verb = {"walks": "walk", "feeds": "feed"}`).
2. Build one shared vocabulary from the source and target sequences
   together.
3. Encode, pad, and shift the sequences (`decoder_inputs`,
   `decoder_labels`).
4. Build an encoder-decoder from `nn.Embedding`, `nn.LSTM` (encoder),
   `nn.LSTM` (decoder), and `nn.Linear`, with `EMBEDDING_SIZE=16` and
   `HIDDEN_SIZE=32`.
5. Train with full teacher forcing for 400 epochs
   (`Adam(lr=0.03)`, gradient clipping `max_norm=1.0`).
6. Generate each target autoregressively and report how many of the 4
   generated sequences exactly match their target.


### Solution

````python
import matplotlib.pyplot as plt

import re
import torch
from torch import nn
import torch.nn.functional as F
from torch.nn.utils.rnn import pack_padded_sequence
SEED = 42

torch.manual_seed(SEED)

# ----- New corpus -----
corpus_p = [
    "Noah walks a dog in the park.",
    "Mia walks a cat in the park.",
    "Emma feeds a bird in the garden.",
    "Liam feeds a fish in the garden.",
]

token_pattern_p = r"[a-z]+|[.,!?;]"
target_token_lists_p = []
for sentence in corpus_p:
    target_token_lists_p.append(re.findall(token_pattern_p, sentence.lower()))

base_verb_p = {"walks": "walk", "feeds": "feed"}

source_token_lists_p = []
for tokens in target_token_lists_p:
    if tokens[2] == "a":
        object_token = tokens[3]
    else:
        object_token = tokens[2]
    source_tokens = [tokens[0], base_verb_p[tokens[1]], object_token, tokens[-2]]
    source_token_lists_p.append(source_tokens)

print("Controlled source-target pairs:")
for source_tokens, target_tokens in zip(source_token_lists_p, target_token_lists_p):
    print("  source:", " ".join(source_tokens))
    print("  target:", " ".join(target_tokens))

# ----- One shared vocabulary -----
special_tokens_p = ["<PAD>", "<UNK>", "<BOS>", "<EOS>"]
PAD_ID_p, UNK_ID_p, BOS_ID_p, EOS_ID_p = 0, 1, 2, 3

token_counts_p = {}
for sequence in source_token_lists_p + target_token_lists_p:
    for token in sequence:
        token_counts_p[token] = token_counts_p.get(token, 0) + 1

ordered_tokens_p = sorted(token_counts_p.keys())
ordered_tokens_p.sort(key=token_counts_p.get, reverse=True)
vocabulary_p = special_tokens_p + ordered_tokens_p

token_to_id_p, id_to_token_p = {}, {}
for token_id, token in enumerate(vocabulary_p):
    token_to_id_p[token] = token_id
    id_to_token_p[token_id] = token

print()
print("Vocabulary size:", len(vocabulary_p))

# ----- Encode, pad, shift -----
source_sequences_p = []
for source_tokens in source_token_lists_p:
    ids = [BOS_ID_p]
    for token in source_tokens:
        ids.append(token_to_id_p.get(token, UNK_ID_p))
    ids.append(EOS_ID_p)
    source_sequences_p.append(ids)

source_lengths_p = torch.tensor([len(s) for s in source_sequences_p], dtype=torch.long)
source_max_length_p = int(source_lengths_p.max().item()) + 1
source_ids_p = torch.full((len(source_sequences_p), source_max_length_p), PAD_ID_p, dtype=torch.long)
for row, sequence in enumerate(source_sequences_p):
    source_ids_p[row, :len(sequence)] = torch.tensor(sequence)

target_sequences_p = []
for target_tokens in target_token_lists_p:
    ids = [BOS_ID_p]
    for token in target_tokens:
        ids.append(token_to_id_p.get(token, UNK_ID_p))
    ids.append(EOS_ID_p)
    target_sequences_p.append(ids)

target_max_length_p = max(len(s) for s in target_sequences_p)
target_ids_p = torch.full((len(target_sequences_p), target_max_length_p), PAD_ID_p, dtype=torch.long)
for row, sequence in enumerate(target_sequences_p):
    target_ids_p[row, :len(sequence)] = torch.tensor(sequence)

decoder_inputs_p = target_ids_p[:, :-1]
decoder_labels_p = target_ids_p[:, 1:]

print("Source IDs shape:", tuple(source_ids_p.shape))
print("Target IDs shape:", tuple(target_ids_p.shape))

# ----- Build the encoder-decoder -----
VOCAB_SIZE_p = len(vocabulary_p)
EMBEDDING_SIZE_p, HIDDEN_SIZE_p = 16, 32

source_embedding_p = nn.Embedding(VOCAB_SIZE_p, EMBEDDING_SIZE_p, padding_idx=PAD_ID_p)
target_embedding_p = nn.Embedding(VOCAB_SIZE_p, EMBEDDING_SIZE_p, padding_idx=PAD_ID_p)
encoder_lstm_p = nn.LSTM(EMBEDDING_SIZE_p, HIDDEN_SIZE_p, batch_first=True)
decoder_lstm_p = nn.LSTM(EMBEDDING_SIZE_p, HIDDEN_SIZE_p, batch_first=True)
output_layer_p = nn.Linear(HIDDEN_SIZE_p, VOCAB_SIZE_p)

parameters_p = []
for module in [source_embedding_p, target_embedding_p, encoder_lstm_p, decoder_lstm_p, output_layer_p]:
    parameters_p += list(module.parameters())

# ----- Train with full teacher forcing -----
optimizer_p = torch.optim.Adam(parameters_p, lr=0.03)
loss_function_p = nn.CrossEntropyLoss(ignore_index=PAD_ID_p)
EPOCHS_p = 400
loss_history_p = []

for epoch in range(1, EPOCHS_p + 1):
    optimizer_p.zero_grad()

    source_vectors = source_embedding_p(source_ids_p)
    packed_source = pack_padded_sequence(
        source_vectors, source_lengths_p.cpu(), batch_first=True, enforce_sorted=False,
    )
    _, (encoder_hidden, encoder_cell) = encoder_lstm_p(packed_source)

    decoder_vectors = target_embedding_p(decoder_inputs_p)
    decoder_outputs, _ = decoder_lstm_p(decoder_vectors, (encoder_hidden, encoder_cell))
    logits = output_layer_p(decoder_outputs)

    loss = loss_function_p(logits.reshape(-1, VOCAB_SIZE_p), decoder_labels_p.reshape(-1))
    loss.backward()
    torch.nn.utils.clip_grad_norm_(parameters_p, max_norm=1.0)
    optimizer_p.step()

    loss_history_p.append(float(loss.item()))
    if epoch == 1 or epoch % 100 == 0:
        print(f"Epoch {epoch:3d} | loss = {loss.item():.6f}")

# ----- Generate autoregressively -----
generated_sequences_p = []
exact_matches_p = []
source_embedding_p.eval()
target_embedding_p.eval()
encoder_lstm_p.eval()
decoder_lstm_p.eval()
output_layer_p.eval()

for row in range(len(source_token_lists_p)):
    with torch.no_grad():
        source_vectors = source_embedding_p(source_ids_p[row:row + 1])
        packed_source = pack_padded_sequence(
            source_vectors, source_lengths_p[row:row + 1].cpu(), batch_first=True, enforce_sorted=False,
        )
        _, (hidden, cell) = encoder_lstm_p(packed_source)
        current_token = torch.tensor([[BOS_ID_p]], dtype=torch.long)
        generated_tokens = []

        for step in range(12):
            current_vector = target_embedding_p(current_token)
            decoder_output, (hidden, cell) = decoder_lstm_p(current_vector, (hidden, cell))
            step_logits = output_layer_p(decoder_output[:, -1])
            next_id = int(step_logits.argmax(dim=-1).item())
            if next_id == EOS_ID_p:
                break
            generated_tokens.append(id_to_token_p[next_id])
            current_token = torch.tensor([[next_id]], dtype=torch.long)

    generated_sequences_p.append(generated_tokens)
    exact_matches_p.append(generated_tokens == target_token_lists_p[row])

    print()
    print("Source   :", " ".join(source_token_lists_p[row]))
    print("Target   :", " ".join(target_token_lists_p[row]))
    print("Generated:", " ".join(generated_tokens))
    print("Exact    :", exact_matches_p[-1])

print()
print("Exact matches:", sum(exact_matches_p), "/", len(exact_matches_p))


# ----- Visualization 1: teacher-forced training loss -----
plt.figure(figsize=(5, 3))
plt.plot(loss_history_p)
plt.xlabel("Epoch")
plt.ylabel("Cross-Entropy Loss")
plt.title("Practice: Seq2Seq Training Loss")
plt.grid(False)
plt.tight_layout()
plt.show()

# ----- Visualization 2: exact match per example -----
example_labels_p = [" ".join(tokens) for tokens in source_token_lists_p]
match_values_p = [int(match) for match in exact_matches_p]

plt.figure(figsize=(5, 3))
plt.bar(range(len(match_values_p)), match_values_p, tick_label=[f"ex {i+1}" for i in range(len(match_values_p))])
plt.ylim(0, 1.2)
plt.ylabel("Exact match (1 = yes)")
plt.title("Practice: Generated vs. Target Match")
plt.grid(False)
plt.tight_layout()
plt.show()
````
