## From low training loss to good generalization

This notebook is a **beginner-level PyTorch exercise**.

- Distinguishes the roles of training, validation, and test sets.
- Compares underfitting, a good fit, and overfitting.
- Reads training loss together with validation loss.
- Practices early stopping, weight decay, and dropout.
- Selects a model using validation data, then evaluates on the test
  set exactly once at the end.
- Computes accuracy, the confusion matrix, precision, recall, and F1.



```python
import torch
from torch import nn
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display
from sklearn.datasets import make_moons
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    ConfusionMatrixDisplay,
    precision_score,
    recall_score,
    f1_score,
)

# Settings so we always get the same results
torch.manual_seed(42)
torch.set_num_threads(1)

print("PyTorch version:", torch.__version__)
```

## Practice Problem

**Scenario:** Repeat the capacity-comparison experiment on a
**different** nonlinear dataset: two concentric circles, generated with
`sklearn.datasets.make_circles(n_samples=600, noise=0.15, factor=0.4,
random_state=42)`.

Using the **same method** as this lecture, do the following:

1. Generate the dataset and split it into training (60%), validation
   (20%), and test (20%) sets, standardizing with the training mean/std.
2. Flip 15% of the training labels the same way as this lecture
   (`torch.Generator().manual_seed(42)`).
3. Build the same three models: `Underfit` (`nn.Linear(2,1)`),
   `Good fit` (`2->16->16->1` with ReLU), and `High capacity`
   (`2->128->128->128->1` with ReLU), with the same learning rates and
   800 epochs.
4. Report the final validation accuracy for all three.
5. Plot the learning curves for the `Good fit` model.


### Solution

```python
from sklearn.datasets import make_circles

# ----- Data -----
X_numpy_p, y_numpy_p = make_circles(n_samples=600, noise=0.15, factor=0.4, random_state=42)

X_train_val_numpy_p, X_test_numpy_p, y_train_val_numpy_p, y_test_numpy_p = train_test_split(
    X_numpy_p, y_numpy_p, test_size=0.20, random_state=42, stratify=y_numpy_p,
)
X_train_numpy_p, X_val_numpy_p, y_train_numpy_p, y_val_numpy_p = train_test_split(
    X_train_val_numpy_p, y_train_val_numpy_p, test_size=0.25, random_state=42, stratify=y_train_val_numpy_p,
)

train_mean_p = X_train_numpy_p.mean(axis=0, keepdims=True)
train_std_p = X_train_numpy_p.std(axis=0, keepdims=True)

X_train_p = torch.tensor((X_train_numpy_p - train_mean_p) / train_std_p, dtype=torch.float32)
X_val_p = torch.tensor((X_val_numpy_p - train_mean_p) / train_std_p, dtype=torch.float32)
X_test_p = torch.tensor((X_test_numpy_p - train_mean_p) / train_std_p, dtype=torch.float32)

y_train_clean_p = torch.tensor(y_train_numpy_p, dtype=torch.float32).reshape(-1, 1)
y_val_p = torch.tensor(y_val_numpy_p, dtype=torch.float32).reshape(-1, 1)
y_test_p = torch.tensor(y_test_numpy_p, dtype=torch.float32).reshape(-1, 1)

# ----- Add label noise, the same way as the lecture -----
y_train_noisy_p = y_train_clean_p.clone()
noise_generator_p = torch.Generator().manual_seed(42)
number_to_flip_p = int(0.15 * len(y_train_noisy_p))
flip_indices_p = torch.randperm(len(y_train_noisy_p), generator=noise_generator_p)[:number_to_flip_p]
y_train_noisy_p[flip_indices_p] = 1 - y_train_noisy_p[flip_indices_p]

print("Flipped labels:", number_to_flip_p, "out of", len(y_train_noisy_p))

# ----- Build the three models -----
torch.manual_seed(42)
underfit_model_p = nn.Sequential(nn.Linear(2, 1))
good_fit_model_p = nn.Sequential(
    nn.Linear(2, 16), nn.ReLU(), nn.Linear(16, 16), nn.ReLU(), nn.Linear(16, 1),
)
high_capacity_model_p = nn.Sequential(
    nn.Linear(2, 128), nn.ReLU(), nn.Linear(128, 128), nn.ReLU(),
    nn.Linear(128, 128), nn.ReLU(), nn.Linear(128, 1),
)

model_configs_p = {
    "Underfit": {"model": underfit_model_p, "learning_rate": 0.02},
    "Good fit": {"model": good_fit_model_p, "learning_rate": 0.02},
    "High capacity": {"model": high_capacity_model_p, "learning_rate": 0.005},
}

loss_fn_p = nn.BCEWithLogitsLoss()
capacity_histories_p = {}

for model_name, config in model_configs_p.items():
    current_model = config["model"]
    optimizer_p = torch.optim.Adam(current_model.parameters(), lr=config["learning_rate"])
    train_loss_history_p, val_loss_history_p = [], []

    for epoch in range(800):
        current_model.train()
        optimizer_p.zero_grad()
        train_logits = current_model(X_train_p)
        train_loss = loss_fn_p(train_logits, y_train_noisy_p)
        train_loss.backward()
        optimizer_p.step()

        current_model.eval()
        with torch.no_grad():
            val_loss = loss_fn_p(current_model(X_val_p), y_val_p)

        train_loss_history_p.append(train_loss.item())
        val_loss_history_p.append(val_loss.item())

    capacity_histories_p[model_name] = {"train_loss": train_loss_history_p, "val_loss": val_loss_history_p}

# ----- Validation accuracy -----
validation_summary_p = []
for model_name, config in model_configs_p.items():
    current_model = config["model"]
    current_model.eval()
    with torch.no_grad():
        val_predictions_p = (torch.sigmoid(current_model(X_val_p)) >= 0.5).to(torch.int64)
    val_accuracy_p = accuracy_score(y_val_p.squeeze().numpy().astype(int), val_predictions_p.squeeze().numpy())
    validation_summary_p.append({"model": model_name, "validation_accuracy": val_accuracy_p})

validation_summary_p = pd.DataFrame(validation_summary_p)
display(validation_summary_p.round(4))

# ----- Learning curve for the Good fit model -----
plt.figure(figsize=(5, 3))
plt.plot(capacity_histories_p["Good fit"]["train_loss"], label="Training loss")
plt.plot(capacity_histories_p["Good fit"]["val_loss"], label="Validation loss")
plt.xlabel("Epoch")
plt.ylabel("BCE loss")
plt.title("Practice: Good-Fit Model on Concentric Circles")
plt.legend()
plt.grid(False)
plt.show()
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

# ----- Decision boundary grid -----
grid_margin_p = 0.5
x_min_p, x_max_p = X_train_p[:, 0].min() - grid_margin_p, X_train_p[:, 0].max() + grid_margin_p
y_min_p, y_max_p = X_train_p[:, 1].min() - grid_margin_p, X_train_p[:, 1].max() + grid_margin_p
xx_p, yy_p = np.meshgrid(
    np.linspace(x_min_p, x_max_p, 300),
    np.linspace(y_min_p, y_max_p, 300),
)
grid_points_p = torch.tensor(
    np.column_stack([xx_p.ravel(), yy_p.ravel()]), dtype=torch.float32
)

fig, axes_p = plt.subplots(1, 3, figsize=(15, 4))
for ax, (model_name, config) in zip(axes_p, model_configs_p.items()):
    current_model = config["model"]
    current_model.eval()
    with torch.no_grad():
        grid_probs_p = torch.sigmoid(current_model(grid_points_p)).reshape(xx_p.shape).numpy()

    ax.contourf(xx_p, yy_p, grid_probs_p, levels=20, cmap="RdBu_r", alpha=0.7, vmin=0, vmax=1)
    ax.contour(xx_p, yy_p, grid_probs_p, levels=[0.5], colors="k", linewidths=1.5)
    ax.scatter(
        X_train_p[:, 0], X_train_p[:, 1],
        c=y_train_noisy_p.squeeze().numpy(),
        cmap="RdBu_r", edgecolors="k", linewidths=0.3, s=15,
    )
    ax.set_title(f"{model_name}")
    ax.set_xlabel("x1")
    ax.set_ylabel("x2")
plt.suptitle("Practice: Decision Boundaries (noisy training labels shown)")
plt.tight_layout()
plt.show()

# ----- Test accuracy + confusion matrices -----
test_summary_p = []
test_predictions_by_model_p = {}
for model_name, config in model_configs_p.items():
    current_model = config["model"]
    current_model.eval()
    with torch.no_grad():
        test_predictions_p = (torch.sigmoid(current_model(X_test_p)) >= 0.5).to(torch.int64)
    test_predictions_by_model_p[model_name] = test_predictions_p.squeeze().numpy()
    test_accuracy_p = accuracy_score(y_test_p.squeeze().numpy().astype(int), test_predictions_by_model_p[model_name])
    test_summary_p.append({"model": model_name, "test_accuracy": test_accuracy_p})

test_summary_p = pd.DataFrame(test_summary_p)
display(test_summary_p.round(4))

fig, axes_p = plt.subplots(1, 3, figsize=(15, 4))
y_test_int_p = y_test_p.squeeze().numpy().astype(int)
for ax, model_name in zip(axes_p, model_configs_p.keys()):
    cm_p = confusion_matrix(y_test_int_p, test_predictions_by_model_p[model_name])
    ConfusionMatrixDisplay(confusion_matrix=cm_p, display_labels=["0", "1"]).plot(ax=ax, colorbar=False)
    ax.set_title(model_name)
    ax.grid(False)
plt.suptitle("Practice: Confusion Matrices on Test Set")
plt.tight_layout()
plt.show()
```
